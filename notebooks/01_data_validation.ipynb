```python
df_clean = df.drop_duplicates().copy()

print("Original rows:", len(df))
print("Rows after removing exact duplicates:", len(df_clean))
print("Rows removed:", len(df) - len(df_clean))
```

```python
march_1_clean = df_clean[
    df_clean["date"] == "01-03-2025"
].copy()

print("Records on 01-03-2025 after duplicate removal:",
      len(march_1_clean))

print("\nRecords per position:")
print(
    march_1_clean["position"]
    .value_counts()
    .sort_index()
)
```

```python
march_1_clean[["position", "song", "artist"]].head(20)
```

```python
print(
    "First position:",
    march_1_clean.iloc[0]["position"]
)

print(
    "Last position:",
    march_1_clean.iloc[49]["position"]
)
```

```python
# Separate the abnormal date from the rest of the dataset
normal_data = df_clean[
    df_clean["date"] != "01-03-2025"
].copy()

# Get the first complete 1–50 snapshot from the abnormal date
march_1_snapshot = (
    march_1_clean
    .iloc[:50]
    .copy()
)

# Combine the cleaned normal data with the retained snapshot
df_clean = pd.concat(
    [normal_data, march_1_snapshot],
    ignore_index=True
)

print("Cleaned dataset shape:", df_clean.shape)
```

```python
daily_counts_clean = df_clean.groupby("date").size()

print(
    "Dates with exactly 50 records:",
    (daily_counts_clean == 50).sum()
)

print(
    "Dates with records other than 50:",
    (daily_counts_clean != 50).sum()
)

print("\nAbnormal dates:")
print(daily_counts_clean[daily_counts_clean != 50])
```

```python
print("Exact duplicate rows remaining:",
      df_clean.duplicated().sum())
```

```python
duplicate_song_date = (
    df_clean.groupby(["date", "song"])
    .size()
    .reset_index(name="count")
)

duplicate_song_date = duplicate_song_date[
    duplicate_song_date["count"] > 1
]

print(
    "Duplicate song-date combinations:",
    len(duplicate_song_date)
)
```

```python
duplicate_song_date
```

```python
print("Number of duplicate song-date combinations:",
      len(duplicate_song_date))

print("\nDates containing duplicate songs:")
print(duplicate_song_date["date"].value_counts())
```

```python
duplicate_song_date.head(12)
```

```python
luna_duplicates = df_clean[
    (df_clean["song"] == "LUNA") &
    (df_clean["date"].isin(
        duplicate_song_date["date"]
    ))
].sort_values(["date", "position"])

luna_duplicates[
    ["date", "position", "song", "artist", "popularity", "album_type"]
]
```

```python
print(
    luna_duplicates.groupby("date")["position"]
    .apply(list)
)
```

```python
duplicate_songs = (
    df_clean.groupby(["date", "song"])
    .size()
    .reset_index(name="count")
)

duplicate_songs = duplicate_songs[
    duplicate_songs["count"] > 1
]

print("Duplicate song-date cases:", len(duplicate_songs))
print("\nUnique songs involved:")
print(duplicate_songs["song"].unique())
```

```python
print(
    duplicate_songs["song"]
    .value_counts()
)
```

```python
print("========== FINAL DATA VALIDATION ==========")

print("Total rows:", len(df_clean))
print("Total columns:", len(df_clean.columns))
print("Unique dates:", df_clean["date"].nunique())

daily_counts_final = df_clean.groupby("date").size()

print("\nDates with exactly 50 records:",
      (daily_counts_final == 50).sum())

print("Dates with records other than 50:",
      (daily_counts_final != 50).sum())

print("\nExact duplicate rows:",
      df_clean.duplicated().sum())

print("\nMissing values:")
print(df_clean.isnull().sum())

print("===========================================")
```

```python
df_clean.to_csv(
    "../data/cleaned_data.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
